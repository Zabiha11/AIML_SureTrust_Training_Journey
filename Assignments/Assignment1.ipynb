Question 1: Clean the Dataset & Find the Most Frequent Category

```python
data = ["red", "blue", "?", "green", "blue", "", "N/A", "green", "blue"]
```

```python
invalid_values = ["?", "N/A", "None", ""]
clean_data = []
```

```python
for item in data:
    if item not in invalid_values:
        clean_data.append(item)
```

```python
freq = {}
```

```python
for item in clean_data:
    if item in freq:
        freq[item] += 1
    else:
        freq[item] = 1
```

```python
most_frequent = None
max_count = 0
```

```python
for key in freq:
    if freq[key] > max_count:
        max_count = freq[key]
        most_frequent = key
```

```python
print("Most Frequent Category:", most_frequent, f"({max_count} times)")
```

Question 2: Compute Moving Average (Window Size K)

```python
temps = [30, 32, 31, 29, 28, 35, 36]
k = 3
moving_avg = []
for i in range(len(temps) - k + 1):
    win_sum = 0
    for j in range(i, i+k):
        win_sum += temps[j]

    avg = win_sum / k
    moving_avg.append(round(avg,2))

print("Moving Averages:", moving_avg)
```
