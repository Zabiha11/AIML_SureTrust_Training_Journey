1. Transaction Frequency Analyzer

```python
def top_freq(transactions):
    freq = {}
    for t in transactions:
        freq[t] = freq.get(t, 0) + 1
    sorted_items = sorted(freq.items(), key=lambda x: (-x[1], x[0]))

    return [item[0] for item in sorted_items[:2]]

transactions = [200, 500, 200, 100, 500, 200, 100, 500, 500]
print(top_freq(transactions))
```

 2. Missing Value Percentage Calculator

```python
data = [10, None, 20, 30, None, None, 40]

total = 0
missing = 0

for value in data:
    total += 1
    if value is None :
        missing += 1

percentage = (missing/total) * 100

print(round(percentage, 2))
```
